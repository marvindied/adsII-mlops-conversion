```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Daten laden
df = pd.read_csv('../data/online_shoppers_intention.csv')
```

```python
# 2. Erster Überblick
print("Datensatz Form:", df.shape)
print("\nFehlende Werte:\n", df.isnull().sum())
```

```python
# 3. Zielvariable analysieren
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Revenue')
plt.title('Verteilung der Käufe (Revenue)')
plt.show()
```

```python
# Ausgabe der exakten Prozentzahlen
print("\nKlassenverteilung in Prozent:\n", df['Revenue'].value_counts(normalize=True) * 100)
```
